```python
import numpy as np
import pandas as pd
```

```python
x = np.array([[1,2], [3,4], [5,6]])
```

```python
print("Array : \n", x)
print("Shape : \n", x.shape)
print("Transpose: \n", x.T)
print("Matrix Multiplication : \n", x@x.T)
print("Matrix Multiplication (Old way): \n", np.dot(x,x.T))
print("Mean of each column : \n", x.mean(axis=0))
print("Mean of each row : \n", x.mean(axis = 1))
print("Random matrix (rows, columns) : \n", np.random.randn(3, 3, 2))
```
